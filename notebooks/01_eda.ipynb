```python
import sys

sys.path.append("..")

import pandas as pd

from src.ingest import detect_language, load_document
from src.settings import DOCS_DIR

rows, texts = [], {}
for p in sorted(DOCS_DIR.rglob("*")):
    if p.is_file() and p.name != ".gitkeep":
        text = load_document(p)
        texts[p] = text
        rows.append({
            "domain": p.parent.name,
            "file": p.name,
            "format": p.suffix,
            "language": detect_language(text),
            "chars": len(text),
        })

df = pd.DataFrame(rows)
print(len(df), "files")
df
```

```python
print(df.groupby("format").size(), "\n")
print(df.groupby("language").size(), "\n")
print(df.groupby("domain").size(), "\n")
df["chars"].describe()
```

```python
ar_pdf = next(p for p in texts if p.suffix == ".pdf" and detect_language(texts[p]) == "ar")
print(ar_pdf.name, "\n")
print(texts[ar_pdf][:500])
```
