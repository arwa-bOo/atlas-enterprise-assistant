```python
lines = EVAL_CASES_PATH.read_text(encoding="utf-8").splitlines()
cases = [json.loads(line) for line in lines if line.strip()]

hits = 0
for case in cases:
    got = {c.source_filename for c in retrieve(case["input"], case["expected_domain"])}
    missing = set(case["expected_sources"]) - got
    hits += not missing
    print(case["id"], case["language"], "OK" if not missing else f"MISS {missing}")
print(f"\n{hits}/{len(cases)} cases fully retrieved")
```
